```python
import gdeltdoc
from gdeltdoc import Filters
from datetime import datetime, timedelta

yesterday = (datetime.now() - timedelta(days=1)).strftime('%Y-%m-%d')
day_before_yesterday = (datetime.now() - timedelta(days=2)).strftime('%Y-%m-%d')

f = Filters(
    start_date = day_before_yesterday,
    end_date = yesterday,
    domain = ['bloomberg.com', 'reuters.com']
    )

gd = gdeltdoc.GdeltDoc()

# Search for articles matching the filters
articles = gd.article_search(f)
```

```python
articles.groupby(['domain']).count()
```

## Entity extraction

```python
filtered_copy = filtered_results.copy()
```

```python
# Run once every time following cells are needed
!python -m spacy download en_core_web_lg
```

```python
import spacy

# Load spaCy NLP model
nlp = spacy.load("en_core_web_lg")

# Predefined topic keywords
TOPIC_KEYWORDS = ["sanctions", "pipeline", "inflation", "OPEC", "trade war", "embargo", "energy crisis"]

def extract_entities(headline):
    """
    Extracts entities from a given news headline.

    :param headline: str, news headline text
    :return: dict containing extracted entities
    """
    doc = nlp(headline)

    # Extract named entities
    locations = [ent.text for ent in doc.ents if ent.label_ == "GPE"]
    organizations = [ent.text for ent in doc.ents if ent.label_ == "ORG"]
    people = [ent.text for ent in doc.ents if ent.label_ == "PERSON"]
    events = [ent.text for ent in doc.ents if ent.label_ == "EVENT"]

    # Extract topics based on keyword matching
    topics = [word for word in TOPIC_KEYWORDS if word.lower() in headline.lower()]

    return {
        "Locations": locations,
        "Organizations": organizations,
        "People": people,
        "Topics": topics,
        "Events": events
    }
```

```python
filtered_copy['Locations'] = None
filtered_copy['Organizations'] = None
filtered_copy['People'] = None
filtered_copy['Topics'] = None
filtered_copy['Events'] = None

for index, row in filtered_copy.iterrows():
  new_row = extract_entities(row['Headline'])
  filtered_copy.at[index, 'Locations'] = new_row['Locations']
  filtered_copy.at[index, 'Organizations'] = new_row['Organizations']
  filtered_copy.at[index, 'People'] = new_row['People']
  filtered_copy.at[index, 'Topics'] = new_row['Topics']
  filtered_copy.at[index, 'Events'] = new_row['Events']
```

### Function to show unique entities and their counts

```python
from collections import Counter

def extract_unique(df, column_name):
    """
    Extract unique names from a column containing lists of names and count occurrences.

    :param df: Pandas DataFrame
    :param column_name: Column containing lists of names
    :return: DataFrame with unique names and their counts
    """
    # Flatten all names from lists into a single list
    all_names = [name for sublist in df[column_name].dropna() for name in sublist]

    # Count occurrences of each unique name
    name_counts = Counter(all_names)

    # Convert to DataFrame
    df_names = pd.DataFrame(name_counts.items(), columns=["Name", "Count"])

    # Sort by most common names
    df_names = df_names.sort_values(by="Count", ascending=False).reset_index(drop=True)

    return df_names
```
